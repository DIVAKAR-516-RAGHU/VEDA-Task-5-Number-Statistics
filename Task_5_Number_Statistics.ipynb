## Task 5 – Number Statistics

### Objective
Create a Python program to identify even and odd numbers and find the largest and smallest numbers in a given list.

### Tools Used
- Python
- Jupyter Notebook

```python
numbers = [12, 7, 25, 4, 18, 9, 30, 11, 6, 21]

even_numbers = []
odd_numbers = []

for number in numbers:
    if number % 2 == 0:
        even_numbers.append(number)
    else:
        odd_numbers.append(number)

largest = max(numbers)
smallest = min(numbers)

print("Original numbers:", numbers)
print("Even numbers:", even_numbers)
print("Odd numbers:", odd_numbers)
print("Largest number:", largest)
print("Smallest number:", smallest)
```

## Explanation

The program uses a list of numbers as input.

- A `for` loop processes each number in the list.
- The modulo operator `%` checks whether a number is divisible by 2.
- If `number % 2 == 0`, the number is even.
- Otherwise, the number is odd.
- `append()` adds numbers to the respective lists.
- `max()` finds the largest number.
- `min()` finds the smallest number.

```python
print("----- Number Statistics -----")
print("Original numbers:", numbers)
print("Even numbers:", even_numbers)
print("Odd numbers:", odd_numbers)
print("Largest number:", largest)
print("Smallest number:", smallest)
```

## Interview Questions

### 1. How do you check whether a number is even?

A number is even if the remainder after dividing it by 2 is zero.

Example:

`number % 2 == 0`

### 2. What does the modulo operator do?

The modulo operator `%` returns the remainder of a division.

Example:

`10 % 3` returns `1`.

### 3. How can loops be used for data processing?

Loops process each element of a collection one by one. In this task, the `for` loop checks every number and separates it into even and odd numbers.
